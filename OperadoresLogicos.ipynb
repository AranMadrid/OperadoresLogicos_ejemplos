```python
# Definición de la proposición original
calor = True

# Negación de la proposición
negacion = not calor

# Imprimir los resultados
print(f"Proposición original, Hace calor: {calor}")
print(f"Negación de la proposición original, Hace calor: {negacion}")
```

conjuncion: "Y" ambas deben ser verdaderas para que toda la oracion sea verdadera

```python
#definicion de dos proposiciones
proposicion1 = True
proposicion2 = False

#conjuncion
conjuncion = proposicion1 and proposicion2

#imprimir los resultados
print (f"Proposicion 1: {proposicion1}")
print (f"Proposicion 2: {proposicion2}")
print (f"Conjuncion: {conjuncion}")
```

disyunción (o) basta con que una sea verdadera para que todo sea verdadero

```python
#definimos proposiciones
proposicion1 = True
proposicion2 = False

disyuncion = proposicion1 or proposicion2

print (f"voy al cine: {proposicion1}")
print (f"me quedo en casa: {proposicion2}")
print (f"voy al cine o me quedo en casa: {disyuncion}")
```

implicacion: (si y entonces) -->

```python
#Ejemplo1: si llueve, entonces llevaré paraguas
llueve = True
llevar_paraguas= llueve #si llueve implica llevar paraguas
print (f"Si llueve: {llueve}")
print (f"Entonces llevaré un paraguas: {llevar_paraguas}")

#Ejemplo2:si estudio entonces aprobaré el examen
estudio = True
aprobar_examen = estudio #si estudio implica aprobar examen
print (f"Si estudio: {estudio}")
print (f"Entonces aprobaré el examen: {aprobar_examen}")

#Ejemplo3:"si como chocolate entonces estaré feliz"
come_chocolate = True
feliz = come_chocolate #si como chocolate implica estar feliz
print (f"Si como chocolate: {come_chocolate}")
print (f"Entonces estaré feliz: {feliz}")
```
